# TASK 6 : Battle-Test Your Model - Support Vector Machines

#### Breakdown Point

The **breakdown point** tells us the level of noise at which the SVM's performance starts to decrease significantly.

In this experiment, the baseline accuracy is approximately **75.37%**. I defined the breakdown point as the first noise level where the accuracy decreases by more than **8 percentage points** from the baseline.

At a noise level of **0.7**:

- Baseline Accuracy = **75.37%**
- Accuracy at 0.7 noise = **66.18%**
- Performance drop = **9.19 percentage points**
- Since 9.19 > 8, the breakdown point is **0.7**.

##### Interpretation

This means that the SVM performs reasonably well with low levels of noise, but when the noise standard deviation reaches **0.7**, the model's performance has degraded significantly according to our chosen threshold.

Therefore, the breakdown point helps us understand the **robustness of the model** and how much feature corruption it can tolerate before its predictions become less reliable.

##### Graph Interpretation

The graph shows that both **Accuracy** and **F1 Score** generally decrease as the noise level increases. The dashed vertical line marks the **breakdown point at 0.7**, where the accuracy has dropped more than 8 percentage points from the baseline.<br>

## Task 8: KNN with Ablation Study

- A **K-Nearest Neighbors (KNN)** classifier was implemented on the Breast Cancer Wisconsin dataset to classify tumors as **malignant (1) or benign (0)**.
- The `id` and completely empty `Unnamed: 32` columns were removed, and the diagnosis labels were encoded numerically.
- The data was divided into **80% training and 20% testing** using stratified sampling.
- Since KNN is distance-based, **StandardScaler** was used to normalize the features. A KNN model with **k = 5** was trained.
- The baseline model achieved **95.61% accuracy** and an **F1-score of 93.83%**, with precision of 97.44% and recall of 90.48%.
- A **feature ablation study** was performed by removing one feature at a time, retraining the model, and comparing accuracy, precision, recall, and F1-score.
- Removing **fractal_dimension_mean** and **concavity_worst** resulted in the lowest accuracy of **93.86%**, indicating that they were among the more influential features in this experiment.
- Correlation analysis showed strong relationships between several features, such as **radius, perimeter, and area**, indicating feature redundancy.
- Overall, the study showed that **KNN performs well with proper feature scaling**, while ablation analysis helps identify influential and redundant features.<br>
