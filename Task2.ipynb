```python
import pandas as pd
import random

data = []

for i in range(1, 301):

    study = round(random.uniform(1, 8), 1)
    attendance = random.randint(50, 100)
    sleep = round(random.uniform(4, 9), 1)
    internet = round(random.uniform(1, 8), 1)

    exam = int(
        study * 10 +
        attendance * 0.5 +
        sleep * 2 -
        internet * 1.5 +
        random.randint(-10, 10)
    )

    exam = max(0, min(100, exam))

    result = "Pass" if exam >= 40 else "Fail"

    data.append([
        i,
        random.randint(18, 25),
        random.choice(["Male", "Female"]),
        study,
        attendance,
        sleep,
        internet,
        random.randint(40, 100),
        exam,
        result
    ])

df = pd.DataFrame(data, columns=[
    "Student_ID",
    "Age",
    "Gender",
    "Study_Hours",
    "Attendance",
    "Sleep_Hours",
    "Internet_Usage",
    "Assignment_Score",
    "Exam_Score",
    "Result"
])

df.to_csv("student_performance_dataset.csv", index=False)

print("Dataset created successfully!")
```
